```python
'''
classe "nave-sucateira": regid, status("atracada", "em missão", "fora do expediente")
    def para alterar status
classe "tarefa": tipo, prioridade, nave designada(regid), status
    def para adicionar nave designada
    def para alterar prioridade
    def para alterar status

lista de naves-sucateiras (adicionar 30 naves, default status="atracada")
lista de tarefas (inicialmente vazia)

def "lista naves":
    def "listar todas"
    def "buscar um regid"
    def "listar por status"

def "processamento de tarefas":
    def "adicionar tarefa"
    def "processar tarefa" (inserir nave na tarefa)
    def "ler tarefas"
'''
```

```python
import random as r
import time as t
'''
tiposnavestatus=["atracada", "no expediente", "fora do expediente"]
tipostarefaprio=range(1,10)
tipostarefastatus=["aberta", "concluída"]

'''
#Classes----------------------------------------------------------------------------------------------------
class NaveSucateira():
    def __init__(self,naveregid):           #regid será numero randomizado, mas precisa garantir que não há repetições
        self.naveregid=naveregid
        self.navestatus="atracada"          #default
        
    def updateNaveStatus(self,navestatus):
        self.navestatus=navestatus
    def printNave(self):
        print(f"Registro da nave: {self.naveregid}\nStatus da nave: {self.navestatus}")
        
class Tarefa():
    def __init__(self, tarefatipo, tarefaprio):
        self.tarefatipo=tarefatipo
        self.tarefaprio=tarefaprio
        self.naveregid=0                            #o regid da nave designada para a tarefa é adicionado depois, valor 0 retorna fallback
        self.tarefastatus="aberta"                  #default
    
    #def updateTarefaPrio(self, tarefaprio):
    #    self.tarefaprio=tarefaprio
    def updateTarefaStatus(self, tarefastatus):
        self.tarefastatus=tarefastatus
    def updateTarefaNave(self, naveregid):
        self.naveregid=naveregid
    def readTarefa(self):
        if self.naveregid==0:
            print(f"Tipo de tarefa: {self.tarefatipo}\nPrioridade: {self.tarefaprio}\nNave designada: nenhuma\nStatus: {self.tarefastatus}")
        elif self.naveregid<10000 and self.naveregid>=1000:
            print(f"Tipo de tarefa: {self.tarefatipo}\nPrioridade: {self.tarefaprio}\nNave designada: {self.naveregid}\nStatus: {self.tarefastatus}")
        else:
            print("Erro no registro de nave desta tarefa!")


#Função de simulação--------------------------------------------------------------------------------------
def testeSimulado():
    
    listaTarefas=[]
    listaHistorico=[]

    #Ações para a lista de naves--------------------------------------------------------------------------
    def criarListaNaves():
        listanaves=[]
        for i in range(30):
            novanave=NaveSucateira(r.randrange(1000,9999))
            if novanave not in listanaves:
                listanaves.append(novanave)
                i+=1
        return listanaves

    def listarTodasNaves(listanaves):
        for i in range(30):
            listanaves[i].printNave()
            i+=1
            
    def buscarRegid(listanaves,naveregid):
        for i in range(30):
            if listanaves[i].naveregid==naveregid:
                listanaves[i].printNave()
                break
            i+=1
                
    def listarPorStatus(listanaves,navestatus):
        for i in range(30):
            if listanaves[i].navestatus==navestatus:
                listanaves[i].printNave()
            i+=1

    def processamentoListaNaves(listanaves):
        for i in range(30):
            if listanaves[i].navestatus!="fora do expediente":
                return listanaves[i].naveregid
            i+=1

    def updateListaNaves(listanaves,naveregid):
        for i in range(30):
            if listanaves[i].naveregid==naveregid and checkHistoricoRegid(naveregid)<5:    #nave entra no expediente
                listanaves[i].updateNaveStatus("no expediente")
                break
            if listanaves[i].naveregid==naveregid and checkHistoricoRegid(naveregid)>=5:   #depois de 5 tarefas, nave sai do expediente
                listanaves[i].updateNaveStatus("fora do expediente")
                break
            i+=1

    #Ações para o histórico de tarefas concluídas------------------------------------------------------------
    def lerHistoricoTarefas():
        if len(listaHistorico)>0:
            for i in listaHistorico:
                listaHistorico[i].readTarefa()
        else:
            print("Histórico vazio.")
            
    def updateHistoricoTarefas(tarefaproc):
        if len(listaHistorico)<=90:
            listaHistorico.insert(0,tarefaproc)
        else:
            listaHistorico.pop()
            listaHistorico.insert(0,tarefaproc)
        return listaHistorico
    
    def checkHistoricoRegid(naveregid):
        q=0
        for i in range(len(listaHistorico)):
            if listaHistorico[i].naveregid==naveregid:
                q+=1
            i+=1
        return q
            

    #Ações para a lista de tarefas----------------------------------------------------------------------------
    def adicionarTarefa():
        tipo=input("Informe o tipo da tarefa: ")
        try:
            prio=int(input("Informe a prioridade da tarefa, de 1 a 10: "))
        except:
            print("Insira um numero inteiro para prioridade.")
            return adicionarTarefa()
        if prio>=1 and prio<=10:
            novaTarefa=Tarefa(tipo,prio)
            listaTarefas.append(novaTarefa)
            sorted(listaTarefas, key=lambda tarefa: tarefa.tarefaprio)      #Toda tarefa adicionada reordena a lista por prioridade          
        else:
            print("Insira um valor de prioridade válido.")
            return adicionarTarefa()
    
    def resolverTarefa(listanaves):
        tarefanaveregid=processamentoListaNaves(listanaves)
        if tarefanaveregid:
            tarefaproc=listaTarefas.pop(0)                                  #Lista já está ordenada
            tarefaproc.updateTarefaNave(tarefanaveregid)
            updateListaNaves(listanaves,tarefanaveregid)
            print(f"Nave {tarefanaveregid} alocada, resolvendo tarefa...")
            t.sleep(3)
            tarefaproc.updateTarefaStatus("concluída")
            print("Tarefa concluída!")
            updateHistoricoTarefas(tarefaproc)
        else:
            t.sleep(1)
            return print("Todas naves estão fora do expediente!")
        
    def lerTarefas():
        if len(listaTarefas)<1:
            print("Não há tarefas inseridas.")
        else:
            for i in listaTarefas:
                listaTarefas[i].readTarefa()
                i+=1
    
    #inicializar lista de naves
    listanavessimul=criarListaNaves()
    
    #Loop de menus--------------------------------------------------------------------------------------------
    def loopMenu0():
        menu0=input("Escolha uma operação: 1 - Consultar naves | 2 - Tarefas | 3 - Terminar teste").strip()
        
        #Menu para consulta de naves
        if menu0=="1":
            def loopMenu1():
                menu1=input("Como deseja consultar? 4 - Listar todas naves | 5 - Buscar por regid | 6 - Listar por status | 7 - Voltar ao menu anterior").strip()
                if menu1=="4":
                    listarTodasNaves(listanavessimul)
                if menu1=="5":
                    try:
                        buscanaveregid=int(input("Digite o regid da nave a procurar: "))
                    except:
                        print("Insira apenas numeros.")
                        return loopMenu1()
                    buscarRegid(listanavessimul,buscanaveregid)
                if menu1=="6":
                    buscanavestatus=input("Escolha o status a procurar: ").strip()
                    listarPorStatus(listanavessimul,buscanavestatus)
                if menu1=="7":
                    return
                return loopMenu1()
            loopMenu1()
            
        #Menu para operações com tarefas 
        if menu0=="2":
            def loopMenu2():
                menu2=input("O que deseja fazer? 4 - Mostrar todas tarefas | 5 - Adicionar uma tarefa | 6 - Resolver uma tarefa | 7 - Mostrar histórico | 8 - Voltar ao menu anterior").strip()
                if menu2=="4":
                    lerTarefas()
                if menu2=="5":
                    adicionarTarefa()
                if menu2=="6":
                    resolverTarefa(listanavessimul)
                if menu2=="7":
                    lerHistoricoTarefas()
                if menu2=="8":
                    return
                return loopMenu2()
            loopMenu2()
        
        #Terminar loop
        if menu0=="3":
            print("Terminando teste.")
            return
        
        return loopMenu0()

    return loopMenu0()

testeSimulado()
```
